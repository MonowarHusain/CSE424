````python
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_classification
from sklearn.preprocessing import StandardScaler

#Dataset with 9 classes
X, y = make_classification(n_samples=1000, n_features=20, n_informative=10,
                           n_redundant=5, n_classes=9, random_state=42)
````

### Data Preprocessing and Splitting
First, I'll scale the features and split the dataset into training and testing sets.

````python
# Scale the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42, stratify=y)

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
````

### Build the Neural Network Model
Now, I'll build a neural network with 4 hidden layers as specified (512, 256, 128, 64 neurons) and an appropriate output layer for 9-class classification using the Adaptive Gradient (AdaGrad) optimizer.

````python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adagrad
from tensorflow.keras.callbacks import EarlyStopping

# Get the number of features and classes
n_features = X_train.shape[1]
n_classes = len(tf.unique(y_train)[0])

# Define the model
model = Sequential([
    Dense(512, activation='relu', input_shape=(n_features,)),
    Dense(256, activation='relu'),
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(n_classes, activation='softmax') # Output layer for 9 classes
])

# Compile the model with AdaGrad optimizer
model.compile(optimizer=Adagrad(),
              loss='sparse_categorical_crossentropy', # Use sparse_categorical_crossentropy for integer labels
              metrics=['accuracy'])

model.summary()
````

### Train the Model
Finally, I'll train the model for 400 epochs with a batch size of 64 and implement Early Stopping to prevent overfitting.

````python
# Define Early Stopping callback
early_stopping = EarlyStopping(monitor='val_loss', patience=20, restore_best_weights=True)

# Train the model
history = model.fit(X_train, y_train,
                    epochs=400,
                    batch_size=64,
                    validation_split=0.2,
                    callbacks=[early_stopping],
                    verbose=1)
````

### Evaluate the Model
Let's evaluate the trained model on the test set.

````python
loss, accuracy = model.evaluate(X_test, y_test, verbose=0)
print(f"Test Loss: {loss:.4f}")
print(f"Test Accuracy: {accuracy:.4f}")
````
